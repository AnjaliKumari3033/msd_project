```python
import os
import pandas as pd
```

```python
DATA_DIR = "../data/raw/aclImdb"

print(os.listdir(DATA_DIR))

TRAIN_DIR = os.path.join(DATA_DIR, "train")
TEST_DIR = os.path.join(DATA_DIR, "test")

print("Contents of dataset folder:")
print(os.listdir(DATA_DIR))

print("\nContents of train folder:")
print(os.listdir(TRAIN_DIR))

print("\nContents of test folder:")
print(os.listdir(TEST_DIR))

OUTPUT_DIR = "../data/processed"
```

```python
def load_reviews(folder_path, label):
    """
    Reads all .txt review files from a folder.

    Parameters:
        folder_path : path to the folder containing reviews
        label       : sentiment label
                      0 = negative
                      1 = positive

    Returns:
        A list of dictionaries containing review text and label.
    """
    
    reviews = []

    for filename in os.listdir(folder_path):

        if filename.endswith(".txt"):

            file_path = os.path.join(folder_path, filename)

            with open(file_path, "r", encoding="utf-8") as file:
                text = file.read()

            reviews.append({
                "text": text,
                "label": label
            })

    return reviews
```

```python
train_pos = load_reviews(
    os.path.join(TRAIN_DIR, "pos"),
    1
)

train_neg = load_reviews(
    os.path.join(TRAIN_DIR, "neg"),
    0
)

print("Positive training reviews:", len(train_pos))
print("Negative training reviews:", len(train_neg))
```

```python
train_data = train_pos + train_neg

print("Total training reviews:", len(train_data))
```

```python
train_df = pd.DataFrame(train_data)

print("Training DataFrame shape:", train_df.shape)
train_df.head()
```

```python
print(train_df["label"].value_counts())
print("Negative reviews:", (train_df["label"] == 0).sum())
print("Positive reviews:", (train_df["label"] == 1).sum())
```

```python
test_pos = load_reviews(
    os.path.join(TEST_DIR, "pos"),
    1
)

test_neg = load_reviews(
    os.path.join(TEST_DIR, "neg"),
    0
)

print("Positive test reviews:", len(test_pos))
print("Negative test reviews:", len(test_neg))
```

```python
test_data = test_pos + test_neg

test_df = pd.DataFrame(test_data)

print("Test DataFrame shape:", test_df.shape)

print(test_df["label"].value_counts())

test_df.head()
```

```python
train_df = train_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

test_df = test_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
train_df.head()
```

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Output folder ready! ✅")
```

```python
train_path = os.path.join(
    OUTPUT_DIR,
    "imdb_train.csv"
)

train_df.to_csv(
    train_path,
    index=False,
    encoding="utf-8"
)

print("Training CSV saved successfully! ✅")
print("Location:", train_path)
```

```python
test_path = os.path.join(
    OUTPUT_DIR,
    "imdb_test.csv"
)

test_df.to_csv(
    test_path,
    index=False,
    encoding="utf-8"
)

print("Testing CSV saved successfully! ✅")
print("Location:", test_path)
```

```python
train_check = pd.read_csv(
    train_path,
    encoding="utf-8"
)

test_check = pd.read_csv(
    test_path,
    encoding="utf-8"
)

print("Training CSV shape:", train_check.shape)
print("Testing CSV shape:", test_check.shape)

print("\nTraining columns:")
print(train_check.columns)

print("\nTesting columns:")
print(test_check.columns)
```
